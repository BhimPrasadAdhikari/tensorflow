**Broadcasting**
It lets you perform operations on tensors that have different shapes.

suppose you want to add a single scalar to a vector containing the multimple values, then broadcasting srteches the single value to the same size of the vector so the one by one addition becomes easier.

tensors can be broadcast if they follow this two simple rules:
1. compare rightmost dimension
2. compatible if:
      - equal
      - one of them is 1

tensor1: (3,1)
tensor2: (1,4)

these two are compatible(I know you don't believe me check it nah!)

**The final shape after broadcasting**
1. For each pair of dimensions, the new dimension will be the larger of the two.
(1, 4) -> 4
(3,1) -> 3
final shape of each tensors is (3,4)

### if one tensor has less number of dimension, 1 is automatically added at the beginnig.
A concept borrowed from the `equivalent feature in numpy`.

under some conditions, smaller tensors are "stretched" automatically to fit larger tensors when running combined operations on them.

Automatic Differentiation & Gradients


```python
# tensorflow remembers what operations were performed on the variables during the forward pass
# during the backward pass, tf traverses this list of operations in reverse order to compute the gradients

x = tf.Variable(4.)

with tf.GradientTape() as tape: # records the operation performed on x in 'tape'
  y = x**2

grad = tape.gradient(y, x)
grad
```

```python
w = tf.Variable(tf.random.normal((3,5)), name='w')
b = tf.Variable(tf.zeros(5, dtype=tf.float32), name='b')
x = [[2., 6., 4.]]

with tf.GradientTape(persistent=True) as tape:
  y = x @ w + b
  loss = tf.reduce_mean(y**2)

print(loss)
# gradient of loss func wrt w and b
[w_grad, b_grad] = tape.gradient(loss, [w, b])
print(b_grad)
my_vars = {
    'w': w,
    'b': b
}

grad = tape.gradient(loss, my_vars)
grad['b']
```

Gradient with respect to a model
all subclasses of `tf.Module` stores variable in trainable_variables

```python
layer = tf.keras.layers.Dense(2, activation='relu')
x = tf.constant([[1.,2.,3.]])

with tf.GradientTape() as tape:
  y = layer(x)
  loss = tf.reduce_mean(y**2)

grad = tape.gradient(loss, layer.trainable_variables)

for var, g in zip(layer.trainable_variables, grad):
  print(var.name)
  print(g)
```

controlling what variable to watch.

```python
x0 = tf.Variable(0.)
x1 = tf.Variable(10.)

with tf.GradientTape(watch_accessed_variables=False) as tape:
  tape.watch(x1) # no gradient is calculated for x0
  y0 = tf.math.sin(x0)
  y1 = tf.nn.softplus(x1)
  y = y0 + y1
  ys = tf.reduce_sum(y)

# dys/dx1 = exp(x1) / (1 + exp(x1)) = sigmoid(x1)
grad = tape.gradient(ys, {'x0': x0, 'x1': x1})

print('dy/dx0:', grad['x0'])
print('dy/dx1:', grad['x1'].numpy())
```

# Intermediate results
calculating gradient with the intermediate values computed inside the tape.

```python
x = tf.constant(3.0)

with tf.GradientTape() as tape:
  tape.watch(x)
  y = x * x
  z = y * y

# Use the tape to compute the gradient of z with respect to the
# intermediate value y.
# dz_dy = 2 * y and y = x ** 2 = 9
print(tape.gradient(z, y).numpy())
```

```python
# Building a modules

class SimpleModule(tf.Module):
  def __init__(self, name=None):
    super().__init__(name=name)
    self.a_variable = tf.Variable(5.0, name='train_me')
    self.non_trainable_variable = tf.Variable(5.0, trainable=False, name='do_not_train_me')
  def __call__(self, x):
    return self.a_variable * x + self.non_trainable_variable

simple_module = SimpleModule(name='simple')

print(simple_module(tf.constant(5.0)))
```

```python
class Dense(tf.Module):
  def __init__(self, in_features, out_features, name=None):
    super().__init__(name=name)
    self.w = tf.Variable(
        tf.random.normal([in_features, out_features]), name='w'
    )
    self.b = tf.Variable(tf.zeros([out_features]), name='b')
  def __call__(self, x):
    y = tf.matmul(x, self.w) + self.b
    return tf.nn.relu(y)


class SequentialModule(tf.Module):
  def __init__(self, name=None):
    super().__init__(name=name)
    self.dense_1 = Dense(3,3)
    self.dense_2 = Dense(3,2)

  def __call__(self, x):
    x = self.dense_1(x)
    return self.dense_2(x)

# a linear model with two dense layers, just made a model !
my_model = SequentialModule(name="the_model")

my_model(tf.constant([[2.0, 2.0, 2.0]]))

# stores variables and tf.module instances
my_model.submodules

# variable
for var in my_model.variables:
  print(var, "\n")
```

```python
class FlexibleDenseModule(tf.Module):
  def __init__(self,out_features, name=None): # input feature is not needed
    super().__init__(name=name)
    self.is_built = False
    self.out_features = out_features
  def __call__(self, x):

    if not self.is_built:
      self.w = tf.Variable(
          tf.random.normal([x.shape[-1], self.out_features]), name='w'
      )
      self.b = tf.Variable(tf.zeros([self.out_features]), name='b')
      self.is_built = True
    y = tf.matmul(x, self.w) + self.b
    return tf.nn.relu(y)


class MySequentialModule(tf.Module):
  def __init__(self, name=None):
    super().__init__(name=name)
    self.dense_1 = FlexibleDenseModule(3)
    self.dense_2 = FlexibleDenseModule(2)

  def __call__(self, x):
    x = self.dense_1(x)
    return self.dense_2(x)

my_model = MySequentialModule(name="my_model")
my_model(tf.constant([[2.0, 2.0, 2.0]]))


```

```python
# Keras layered model
class FlexibleDense(tf.keras.layers.Layer):
  # Note the added `**kwargs`, as Keras supports many arguments
  def __init__(self, out_features, **kwargs):
    super().__init__(**kwargs)
    self.out_features = out_features

  def build(self, input_shape):  # Create the state of the layer (weights)
    self.w = tf.Variable(
      tf.random.normal([input_shape[-1], self.out_features]), name='w')
    self.b = tf.Variable(tf.zeros([self.out_features]), name='b')

  def call(self, inputs):  # Defines the computation from inputs to outputs
    return tf.matmul(inputs, self.w) + self.b

# Create the instance of the layer
flexible_dense = FlexibleDense(out_features=3)
  # as its convenient to create variable after being sure of the input shape
  # build is called exactly once and it is called with the shape of the input

```

```python
# keras.Model inherits Layers and comes with extra functionality
# @tf.keras.saving.register_keras_serializable()
class MySequentialModel(tf.keras.Model):
  def __init__(self, name=None, **kwargs):
    super().__init__(**kwargs)

    self.dense_1 = FlexibleDense(out_features=3)
    self.dense_2 = FlexibleDense(out_features=2)

  def call(self, x):

    x = self.dense_1(x)
    return self.dense_2(x)

my_sequential_model = MySequentialModel(name="the_model")

my_sequential_model(tf.constant([[2.0, 2.0, 2.0]]))

my_sequential_model.summary()
```
